```python
import numpy as np
import pandas as pd

# Base de vendas
dados_vendas = {
    'cliente_id': [101, 102, 103, 101, 104, 102, 105, 103],
    'valor': [3500.75, 189.50, np.nan, 1200.00, 450.00, np.nan, 89.90, 780.50],
    'categoria': [
        'Eletronicos',
        'Livros',
        'Roupas',
        'Eletronicos',
        'Automotivo',
        'Livros',
        'Roupas',
        'Roupas',
    ],
    'data_hora': [
        '2024-01-15 10:23:00',
        '2024-01-18 14:05:00',
        '2024-02-05 09:12:00',
        '2024-02-20 16:40:00',
        '2024-03-02 11:00:00',
        '2024-03-15 18:30:00',
        '2024-04-10 08:20:00',
        '2024-04-22 13:45:00',
    ],
    'status': [
        'Concluído',
        'Concluído',
        'Pendente',
        'Concluído',
        'Cancelado',
        'Concluído',
        'Concluído',
        'Concluído',
    ],
    'email': [
        'maria@gmail.com',
        'joao@outlook.com',
        'ana@yahoo.com',
        'maria@gmail.com',
        'carlos@gmail.com',
        'joao@outlook.com',
        'lucas@empresa.com.br',
        'ana@yahoo.com',
    ],
}

# Base de clientes
dados_clientes = {
    'cliente_id': [101, 102, 103, 104, 105],
    'nome': [
        'Maria Silva',
        'Joao Souza',
        'Ana Oliveira',
        'Carlos Lima',
        'Lucas Mendes',
    ],
    'cidade': [
        'Sao Paulo',
        'Rio de Janeiro',
        'Belo Horizonte',
        'Curitiba',
        'Salvador',
    ],
}

df_vendas = pd.DataFrame(dados_vendas)
df_clientes = pd.DataFrame(dados_clientes)
```

# **Parte 1: Diagnóstico e Limpeza (Básico)**

```python
print('1. Dimensões e resumo dos tipos de dados de df_vendas:')
print(df_vendas.shape)
df_vendas.info()

print('\n2. Tratamento de valores nulos na coluna "valor":')
print('Valores nulos antes do preenchimento:\n', df_vendas['valor'].isnull().sum())

# Calcular a mediana global do valor
mediana_global_valor = df_vendas['valor'].median()

# Preencher valores nulos em 'valor' com a mediana da categoria, se disponível, senão com a mediana global
df_vendas['valor'] = df_vendas.groupby('categoria')['valor'].transform(lambda x: x.fillna(x.median()))

# Preencher quaisquer valores nulos restantes com a mediana global (caso alguma categoria inteira fosse NaN)
df_vendas['valor'] = df_vendas['valor'].fillna(mediana_global_valor)

print('Valores nulos depois do preenchimento:\n', df_vendas['valor'].isnull().sum())
print('DataFrame df_vendas após preenchimento de nulos no valor:\n', df_vendas)

print('\n3. Transações com status "Concluído" e valor superior a R$ 500,00:')
transacoes_filtradas = df_vendas.loc[(df_vendas['status'] == 'Concluído') & (df_vendas['valor'] > 500.00)]
print(transacoes_filtradas)
```

# **Cruzamento e Transformação (Intermediário)**

```python
print('4. Junção relacional (left merge) entre df_vendas e df_clientes:')
df_merged = pd.merge(df_vendas, df_clientes, on='cliente_id', how='left')
print(df_merged)

print('\n5. Nova coluna "media_categoria" com valor médio de vendas por categoria:')
df_merged['media_categoria'] = df_merged.groupby('categoria')['valor'].transform('mean')
print(df_merged[['cliente_id', 'categoria', 'valor', 'media_categoria']])

print('\n6. Contagem de clientes que usam @gmail.com:')
gmail_clients_count = df_merged['email'].str.contains('@gmail.com', na=False).sum()
print(f'Número de clientes com e-mail @gmail.com: {gmail_clients_count}')
```

# **Parte 3: Análise Temporal e Agregação (Intermediário)**

```python
print('7. Conversão de data_hora para datetime e extração do mês:')
df_merged['data_hora'] = pd.to_datetime(df_merged['data_hora'])
df_merged['mes_venda'] = df_merged['data_hora'].dt.strftime('%B') # Extrai o nome completo do mês
# Ou para o nome do dia da semana: df_merged['dia_semana'] = df_merged['data_hora'].dt.day_name()
print(df_merged[['data_hora', 'mes_venda']].head())

print('\n8. Tabela dinâmica de valor de vendas por categoria e cidade:')
pivot_table_vendas = pd.pivot_table(df_merged, values='valor', index='categoria', columns='cidade', aggfunc='sum', fill_value=0)
print(pivot_table_vendas)
```
